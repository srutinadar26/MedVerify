```python
import os

healthstory_path = "../datasets/raw/training/HealthStory"

for root, dirs, files in os.walk(healthstory_path):
    for file in files:
        if "review" in file.lower():
            print(os.path.join(root, file))
```

```python
import os

keywords = ["fake", "real", "true", "false", "label", "rating", "review"]

for root, dirs, files in os.walk(healthstory_path):
    for file in files:
        if file.endswith(".json"):
            file_path = os.path.join(root, file)

            try:
                with open(file_path, "r", encoding="utf-8") as f:
                    content = f.read().lower()

                found = [word for word in keywords if word in content]

                if found:
                    print(file, "→", found)

            except Exception as e:
                print("Could not read:", file, e)
```

```python
import pandas as pd
import os

coaid_path = "../datasets/raw/training/CoAID"

for root, dirs, files in os.walk(coaid_path):
    for file in files:
        if file.endswith(".csv"):
            path = os.path.join(root, file)

            df = pd.read_csv(path)

            print("\nFILE:", file)
            print("COLUMNS:")
            print(df.columns.tolist())
            print("SHAPE:", df.shape)
            print()
            
            print(df.head(2).to_string())
            
            break
```

```python
import os
import pandas as pd

coaid_path = "../datasets/raw/training/CoAID"

for root, dirs, files in os.walk(coaid_path):

    for file in files:

        if "NewsFake" in file or "NewsReal" in file:

            path = os.path.join(root, file)

            try:
                df = pd.read_csv(path)

                print("\n==============================")
                print("FILE:", file)
                print("SHAPE:", df.shape)
                print("COLUMNS:", df.columns.tolist())
                print(df.head(2).to_string())

            except Exception as e:
                print("ERROR:", e)
```
